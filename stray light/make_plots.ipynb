```python
%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from datetime import datetime
from matplotlib import dates as mdates
```

```python
df = pd.read_csv('/home/ulyanov/data/solo/phi/wcs/fdt/wcs.csv', skipinitialspace=True).dropna()
dsun = df['dsun_au'].to_numpy()
```

```python
df = pd.read_csv('/home/ulyanov/data/solo/phi/wcs/fdt/cld_fit.csv', skipinitialspace=True).dropna()
dates = np.array([datetime.fromisoformat(date) for date in df['date']])

alpha = df['alpha'].to_numpy()
beta = df['beta'].to_numpy()
epsilon = df['epsilon'].to_numpy()
scale = df['scale'].to_numpy()
bias = df['bias'].to_numpy()
drsun = df['drsun'].to_numpy()
sigma = df['sigma'].to_numpy()
```

```python
plt.rcParams['date.converter'] = 'concise'

fig, ax = plt.subplots(figsize=(10,8))
ax.plot(dates, alpha, color='black', lw=1)
ax.plot([dates[0], dates[-1]], [np.mean(alpha), np.mean(alpha)], color='red', lw=2)
ax.set_xlim(datetime(2023,12,31), datetime(2026,1,1))

ax.set_ylabel(r'$\alpha$', size=14)
ax.grid(True, axis='y')
plt.tight_layout()
```
